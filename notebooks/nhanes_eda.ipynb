```python
data = pd.read_csv(r"C:\Users\amyud\OneDrive\Documents\NHANES_Mental_Health_Project\clean_data\merged_data.csv", encoding='utf-8')
```

```python
data.drop(columns = 'Unnamed: 0', inplace = True)
data.set_index('SEQN', inplace =True)
data.rename(columns={'DPQ_TOTAL': 'PHQ-9_SCORE'}, inplace=True)
data
```

## Age Variable

```python
data[['RIDAGEYR']].describe()
```

Our data show the age variable has a mean age of 52 with an SD of 18.5. The minimum age is 18 because we have filtered the data to exclude participants under 18, and the maximum age documented is 80.
Please note that, via NHANES, all participants aged 80 and older were recorded as 80.

```python
data.groupby('AGE_GROUP').agg({'PHQ-9_SCORE':['mean', 'median', 'std']}).reset_index()
```

This chart shows us basic information about the mean PHQ-9 score across the various age groups. The mean PHQ-9 score for 18-34 shows a mean of mild depression, while the other age groups show a mean of minimal to no depression. The mean PHQ-9 score decreases with age.

### Relationship with PH9 and Age Groups

```python
#order the ages from youngest to oldest
age_order = ['18-34', '35-49', '50-64', '65+']

plt.figure(figsize=(9, 5))
ax = sns.barplot(x='AGE_GROUP', y='PHQ-9_SCORE', data=data, order=age_order, errorbar=None)
plt.xlabel('Age Groups')
plt.ylabel('PHQ-9 Scores')
plt.title('Mean Depression Severity Scores across the Age Groups')
plt.yticks(range(0, 8, 1))
ax.bar_label(ax.containers[0], fmt='%.2f')
plt.show()


#severity
plt.figure(figsize=(9, 5))
ax = sns.boxplot(x='AGE_GROUP', y='PHQ-9_SCORE', data=data, order=age_order)
plt.xlabel('Age Groups')
plt.ylabel('PHQ-9 Scores')
plt.title('Severity of Depression Across the Age Groups')
plt.show()
```

The 50th percentile values for all age groups sit below 5, and the high extreme quartile for the 18-34 age group sits above 15, while the other age groups sit below 15. All age groups show a significant number of outliers, with the 65+ age group showing the greatest outliers (we will investigate that later)

#### Prevalence of Elevated Depression Across Age Groups

```python
# Create a yes/no(1,0) column for participants who meet the threshold for moderate to severe depression
data['MODERATE_TO_SEVERE'] = (data['PHQ-9_SCORE'] >=10).astype(int)


plt.figure(figsize=(9, 5))
ax = sns.barplot(x='AGE_GROUP', y='MODERATE_TO_SEVERE', data=data, order=age_order, errorbar=None)
plt.xlabel('Age Groups')
plt.ylabel('Prevalance(%)')
plt.title('Prevalence of Moderate to Severe Depression across the Age Groups')

#set y and labels to percentage
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0%}'))
ax.bar_label(ax.containers[0], labels=[f'{v.get_height():.1%}' for v in ax.containers[0]])
plt.show()
```

About 13.6% of participants aged 18-34 were recorded to have moderate to severe depression. Age groups 35-49 and 50-64 have a similar prevalence of around 8%, while the 65+ group shows a prevalence of 6.2%, significantly lower than the 18-34 age group.

#### Distribution of Depression Across Age Groups

```python
plt.figure(figsize=(9, 5))
sns.kdeplot(data=data, x='PHQ-9_SCORE', hue='AGE_GROUP', hue_order=age_order, cut=0)
plt.xlabel('Age Group')
plt.ylabel('Density')
plt.title('Depression Distribution by Age Group')
plt.show()
```

## Race Variable

```python
data.groupby('RIDRETH3').agg({'PHQ-9_SCORE': ['count','mean', 'std']})
```

### PHQ-9 Severity by Race

```python
severity_order = ['Minimal', 'Mild', 'Moderate', 'Moderately severe', 'Severe']
severity_table = pd.crosstab(
    data['RIDRETH3'], 
    data['DPQ_SEVERITY'], 
    normalize='index'
)[severity_order]* 100


plt.figure(figsize=(10, 8))
ax = severity_table.plot(kind='bar', stacked=True, colormap='Blues', edgecolor='blue')

plt.title('Depression Severity Distribution Across Races')
plt.xlabel('Race Group')
plt.ylabel('Percentage (%)')
plt.legend(title='Severity Level',bbox_to_anchor=(1.05, 1), loc='upper left')
```

All races show approximately 60% of minimal depression, with Other Hispanic and Other Race/Multiracial showing the greatest percentage of participants with Mild to Severe Depression and Asian/Non Hispanic groups showing the lowest.

## Income Variable

```python
data.groupby('INDFMPIR').agg({'PHQ-9_SCORE': ['count',lambda x: (x.count() / len(data)) * 100,'mean', 'std']})
```

This chart gives us information on the Ratio of Family Income to Poverty. A ratio below 1 shows us that the family's income is below the official poverty threshold; a ratio of 1 shows us that the family's income is at the official poverty threshold. A ratio of 2 shows us that the family's income is above the official poverty threshold(200%). It caps at a ratio of 5, so families whose income is higher than 5 times the official poverty threshold are grouped as 5. 
The mean PHQ-9 score decreases as the income-to-poverty ratio increases. 

### PHQ-9 Severity and Prevalence by Income

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(x='INDFMPIR', y='PHQ-9_SCORE', data=data, errorbar=None)
plt.xlabel('Race')
plt.ylabel('PHQ-9 Scores')
plt.title('Mean Depression Scores Across All Income Class')
plt.yticks(range(0, 8, 1))
ax.bar_label(ax.containers[0], fmt='%.2f')
plt.show()

#Severity of depression

plt.figure(figsize=(10, 5))
ax = sns.boxplot(x='INDFMPIR', y='PHQ-9_SCORE', data=data)
plt.xlabel('Race')
plt.ylabel('PHQ-9 Scores')
plt.title('Severity of Depression Across All Income Class')
plt.show()
```

All the income classes have the 50th percentile below 5 and a significant number of outliers. The 5.0 ratio group has the 75th percentile below 5 as well. The 1.0 ratio group has the 75th percentile above 20. 

#### Prevalence of Elevated Depression Across Income Classes

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(x='INDFMPIR', y='MODERATE_TO_SEVERE', data=data, errorbar=None)
plt.xlabel('Income Class')
plt.ylabel('Prevalance(%)')
plt.title('Prevalence of Moderate to Severe Depression Across the Income Class')

#set y and labels to percentage
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0%}'))
ax.bar_label(ax.containers[0], labels=[f'{v.get_height():.1%}' for v in ax.containers[0]])
plt.show()
```

While the mean of the PHQ-9 score sees a consistent decrease across the income class, the prevalence of moderate to severe depression in the 1.0 ratio sees a 2.1% spike from the less than 1.0 ratio group, and then gradually decreases afterwards.

## Corelation Table

```python
corr_columns = ['RIDAGEYR', 'BMXBMI', 'PHQ-9_SCORE', 'SLQ_AVERAGE', 'PAD_SA/W', 'TOTAL_MV/W', 'INDFMPIR']

# calculate the correlation matrix
corr_matrix = data[corr_columns].corr()

# rename columns/index for readability
rename = ['Age', 'BMI', 'PHQ-9 Score', 'Sleep Hours', 'Sedentary Min', 'Physical Activity', 'Income/PIR']
corr_matrix.columns = rename
corr_matrix.index = rename

# plot the correlation matrix as a heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Demographic Variables')
plt.show()
```

## Missingness Report

```python
dfs = {'RIAGENDR': data['RIAGENDR'], 'RIDAGEYR': data['RIDAGEYR'], 'RIDRETH3': data['RIDRETH3'], 
       'DMDEDUC2': data['DMDEDUC2'], 'INDFMPIR': data['INDFMPIR'], 'BMXBMI': data['BMXBMI'],
       'SLQ_CATEGORY_AVG': data['SLQ_CATEGORY_AVG'], 'TOTAL_MV/W': data['TOTAL_MV/W'], 
       'PAD_SA/W': data['PAD_SA/W']}

dcells = []
for i, j in dfs.items():
    dcells.append({
        'Variable': i,
        'Total Respondents': len(j),
        'Valid count': round(j.count()),
        'Missing count': round(j.isna().sum()),
        'Valid %': round(j.count()/ len(j) * 100, 2),
        'Missing %': round(j.isna().mean().mean() * 100, 2)
        })

variables_info = pd.DataFrame(dcells)
variables_info
```
