## Cleaning/Analyzing Depresssion Scale Data

```python
dpq = dpq.round(0)
dpq
```

```python
#drop all empty rows

dpq_cols = ['DPQ010', 'DPQ020', 'DPQ030', 'DPQ040', 'DPQ050', 'DPQ060', 'DPQ070', 'DPQ080', 'DPQ090']
dpq = dpq.dropna(how = 'all', subset=dpq_cols)
```

```python
# replace missing 

dpq = dpq.replace([7,9], np.nan)
```

```python
#create a total category

# Count how many valid answers each person provided
valid = dpq[dpq_cols].notna().sum(axis=1)

# Calculate mean per item and scale back to a 9-item total if at least 7 items were answered
mean_score = dpq[dpq_cols].mean(axis=1)
dpq['DPQ_TOTAL'] = np.where(valid >= 7, mean_score * 9, np.nan)
```

```python
dpq[dpq['DPQ_TOTAL'] == 9]
```

```python
#create category bins

dpq_bins = [-1, 4, 9, 14, 19, 27]
dpq_labels = ['Minimal', 'Mild', 'Moderate', 'Moderately severe', 'Severe']

dpq['DPQ_SEVERITY'] = pd.cut(dpq['DPQ_TOTAL'], bins=dpq_bins, labels=dpq_labels, right=True)
```

```python
# categorise all data values

dpq[dpq_cols] = dpq[dpq_cols].replace(0, 'Not at all')
dpq[dpq_cols] = dpq[dpq_cols].replace(1, 'Several days')
dpq[dpq_cols] = dpq[dpq_cols].replace(2, 'More than half the days')
dpq[dpq_cols] = dpq[dpq_cols].replace(3, 'Nearly every day')
```

```python
# categorise DPQ100, put NaN as 'Not difficult at all

dpq['DPQ100'] = dpq['DPQ100'].replace(0, 'Not difficult at all')
dpq['DPQ100'] = dpq['DPQ100'].replace(1, 'Somewhat difficult')
dpq['DPQ100'] = dpq['DPQ100'].replace(2, 'Very difficult')
dpq['DPQ100'] = dpq['DPQ100'].replace(3, 'Extremely difficult')
dpq['DPQ100'] = dpq['DPQ100'].replace(np.nan, 'Not difficult at all')
```

## Cleaning/Analyzing Physical Activity Data

```python
paq = paq.round(0)
```

```python
paq1 = paq
```

```python
# replace missing 

paq1 = paq1.replace([7777,9999], np.nan)
```

```python
#drop all empty rows

paq1.columns
paq1_cols =['PAD790Q', 'PAD790U', 'PAD800', 'PAD810Q', 'PAD810U', 'PAD820',
       'PAD680']

paq1= paq1.dropna(how ='all', subset = paq1_cols)
```

```python
#convert all PAD790Q to weeks
paq1.loc[paq1['PAD790U'] == 'D', 'PAD790Q'] *= 7
paq1.loc[paq1['PAD790U'] == 'M', 'PAD790Q'] /= 4
paq1.loc[paq1['PAD790U'] == 'Y', 'PAD790Q'] /= 52

paq1['PAD790U'] = paq1['PAD790U'].str.replace('D','W')
paq1['PAD790U'] = paq1['PAD790U'].str.replace('M','W')
paq1['PAD790U'] = paq1['PAD790U'].str.replace('Y','W')
```

```python
#convert all PAD810Q to weeks
paq1.loc[paq1['PAD810U'] == 'D', 'PAD810Q'] *= 7
paq1.loc[paq1['PAD810U'] == 'M', 'PAD810Q'] /= 4
paq1.loc[paq1['PAD810U'] == 'Y', 'PAD810Q'] /= 52

paq1['PAD810U'] = paq1['PAD810U'].str.replace('D','W')
paq1['PAD810U'] = paq1['PAD810U'].str.replace('M','W')
paq1['PAD810U'] = paq1['PAD810U'].str.replace('Y','W')
```

```python
#convert minutes of moderate, vigorous and sedentary activity to weekly minutes
paq1['PAD_MA/W'] = paq1['PAD800'] * paq1['PAD790Q']
paq1['PAD_VA/W'] = paq1['PAD820'] * paq1['PAD810Q']
paq1['PAD_SA/W'] = paq1['PAD680'] * 7
```

```python
#fill missing values with 0 for active minutes
paq1['PAD_MA/W'] = paq1['PAD_MA/W'].fillna(0)
paq1['PAD_VA/W'] = paq1['PAD_VA/W'].fillna(0)
```

```python
#calculate total activity per week, double the PAD_VA/W
paq1['TOTAL_MV/W'] = paq1['PAD_MA/W'] + (paq1['PAD_VA/W'] * 2)
```

```python
#create category bins
bins = [-1, 150, 300, float('inf')]
labels = ['Low activity', 'Moderate activity', 'High activity']

paq1['PHY_ACTIVITY_CAT'] = pd.cut(paq1['TOTAL_MV/W'], bins=bins, labels=labels)
```

```python
paq = paq1
paq
```

## Cleaning/Analyzing Sleep Data

```python
slq1 = slq
```

```python
# replace missing 

slq1 = slq1.replace([7777,9999], np.nan)
```

```python
# drop all empty columns

slq1.columns
slq1_cols = ['SLQ300', 'SLQ310', 'SLD012', 'SLQ320', 'SLQ330', 'SLD013']

slq1 = slq1.dropna(how = 'all', subset= slq1_cols)
```

```python
# find the average sleep time
slq1['SLQ_AVERAGE'] = (slq1['SLD012'] + slq1['SLD013']) / 2
```

```python
# create a sleep time category

slq_bins = [0, 3, 5, 7, 9, 14, np.inf]
slq_labels = ['Less than 3 hours', '3-5 hours', '5-7 hours', '7-9 hours', '9-14 hours', '14 or more hours']

# Create sleep time category for weekdays
slq1['SLQ_CATEGORY_WD'] = pd.cut(slq1['SLD012'], bins=slq_bins, labels=slq_labels, right=False)

# Create sleep time category for weekends
slq1['SLQ_CATEGORY_WE'] = pd.cut(slq1['SLD013'], bins=slq_bins, labels=slq_labels, right=False)

# Create sleep time category for slq_average
slq1['SLQ_CATEGORY_AVG'] = pd.cut(slq1['SLQ_AVERAGE'], bins=slq_bins, labels=slq_labels, right=False)
```
